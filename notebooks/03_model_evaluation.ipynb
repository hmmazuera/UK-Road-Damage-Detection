```python
from ultralytics import YOLO
```

Load trained model

```python
model = YOLO('/content/drive/MyDrive/UK-Road-Damage-Detection/models/road_damage_yolov8n_100/weights/best.pt')
```

Evaluate on Test Set

Measure precision, recall, mAP@50 and mAP@50-95 on the 499 test images.

These images were not used for training or model selection.

```python
metrics = model.val(
    data='/content/drive/MyDrive/UK-Road-Damage-Detection/data.yaml',
    split='test',
    project='/content/drive/MyDrive/UK-Road-Damage-Detection/models',
    name='test_evaluation'
)
```

Test Results

The selected model was evaluated on 499 test images containing 1,048 annotated objects.

- Precision: 0.672.
- Recall: 0.558.
- mAP@50: 0.599.
- mAP@50-95: 0.278.

Test mAP remained close to validation performance.
Lateral Crack had the lowest recall, at 0.364.

Evaluation ran on CPU, with an inference time of 130.1 ms per image in this environment.

Inspect Predictions

```python
from pathlib import Path
```

```python
images = sorted(Path("/content/drive/MyDrive/UK-Road-Damage-Detection/data/test/images").glob('*.jpeg'))

results = model.predict(images[1], conf=0.25)
results[0].show()
```

Evaluation Summary

- Evaluated the selected model on 499 test images.
- Achieved precision of 0.672 and recall of 0.558.
- Achieved mAP@50 of 0.599 and mAP@50-95 of 0.278.
- Inspected predictions using a confidence threshold of 0.25.
- No detections were returned for AA_10800.jpeg.
- One Longitudinal Crack was detected in AB_12560.jpeg.
- Lateral Crack remained the weakest class.

The model can miss road defects, so an image without detections should not be interpreted as damage-free.
