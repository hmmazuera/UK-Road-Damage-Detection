Objectives:
- Load the prepared dataset.
- Train YOLOv8n using transfer learning.
- Save the trained model for evaluation and deployment.

```python
from ultralytics import YOLO
```

```python
model = YOLO('yolov8n.pt')
```

Training model using transfer learning

```python
results = model.train(
    data='/content/drive/MyDrive/UK-Road-Damage-Detection/data.yaml',
    epochs=50,
    imgsz=640,
    batch=16,
    patience=10,
    workers=2,
    project='/content/drive/MyDrive/UK-Road-Damage-Detection/models',
    name='road_damage_yolov8n'
)
```

Train for 100 Epochs

Train YOLOv8n from the same pretrained weights for up to 100 epochs.

Keep the dataset and other training settings unchanged. Save this experiment separately to compare its validation results with the 50-epoch model.

```python
results = model.train(
    data="/content/drive/MyDrive/UK-Road-Damage-Detection/data.yaml",
    epochs=100,
    imgsz=640,
    batch=16,
    patience=10,
    workers=2,
    project="/content/drive/MyDrive/UK-Road-Damage-Detection/models",
    name="road_damage_yolov8n_100"
)
```

Training Comparison

Two YOLOv8n models were trained using the same dataset split, with maximum training lengths of 50 and 100 epochs.

The 100-epoch experiment achieved better overall validation results:
- Precision: 0.651.
- Recall: 0.588.
- mAP@50: 0.604.
- mAP@50-95: 0.276.

The saved weights remained approximately 6.5 MB.

The best.pt checkpoint from the 100-epoch experiment was selected for test evaluation. Lateral Crack remained the weakest class.
